```python
#Missing Values Table
missing_data = pd.DataFrame({
    "Missing_Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

print(missing_data)
```

```python
#Missing Values + Duplicate Check

# Missing Values
print("Missing Values Before Cleaning:")
print(df.isnull().sum())

# Handle Missing Values
df["Discount_Percent"] = df["Discount_Percent"].fillna(
    df["Discount_Percent"].median()
)

df["Payment_Method"] = df["Payment_Method"].fillna(
    df["Payment_Method"].mode()[0]
)

df["Customer_Type"] = df["Customer_Type"].fillna(
    df["Customer_Type"].mode()[0]
)

# Duplicate Records
print("\nDuplicate Records:", df.duplicated().sum())

# Remove Duplicates
df = df.drop_duplicates()

# Final Check
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDataset Shape:", df.shape)
```

```python
#Convert Date into Datetime
df["Date"] = pd.to_datetime(df["Date"])

#check data type
print(df["Date"].dtype)

#check first 5 dates
print(df["Date"].head())
```

```python
#NumPy Numerical Analysis
sales = df["Net_Sales"].to_numpy()

print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Minimum Sales:", np.min(sales))
print("Maximum Sales:", np.max(sales))
```

```python
#Sales Analysis by City, Category, Product, Salesperson & Customer Type

#Sales by City
print("city-wise Sales:")
print(df.groupby("City")["Net_Sales"].sum().sort_values(ascending=False))
```

```python
#Sales by Category
print("Category-wise Sales:")
print(df.groupby("Category")["Net_Sales"].sum().sort_values(ascending=False))
```

```python
#Sales by Product

print("Product wise- Sales:")
print(df.groupby("Product")["Net_Sales"].sum().sort_values(ascending=False))
```

```python
#Sales by salesperson
print("Salesperson-wise Sales:")
print(df.groupby("Salesperson")["Net_Sales"].sum().sort_values(ascending=False))
```

```python
#Sales by customer Type
print("Customer Type-wise Sales:")
print(df.groupby("Customer_Type")["Net_Sales"].sum().sort_values(ascending=False))
```

```python
#Sales Analysis by Month, Quarter & Year
print("Month-wise Sales:")
print(df.groupby(["Month_Number", "Month"])["Net_Sales"].sum().sort_index())
```

```python
#Quarterly sales
print("Quarter-wise Sales:")
print(df.groupby("Quarter")["Net_Sales"].sum())
```

```python
#Yearly Sales
print("Year-wise Sales:")
print(df.groupby("Year")["Net_Sales"].sum())
```

```python
 #Filtering & Sorting
# FIlter: sales greater than 50,000
print("Net Sales > 50000:")
print(df[df["Net_Sales"] > 50000])
```

```python
#sort:Highest Net Sales First
print("Highest Sales:")
print(df.sort_values("Net_Sales", ascending=False).head(10))
```

```python
#sort: Lowest Net Sales first
print("Lowest Sales:")
print(df.sort_values("Net_Sales").head(10))
```

```python
#City-wise Sales: Bar Chart
city_sales = df.groupby("City")["Net_Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(4,2))
plt.bar(city_sales.index, city_sales.values)

plt.title("City-wise Sales")
plt.xlabel("City")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
#Monthly Sales Trend: Line Chart
monthly_sales = df.groupby(["Month_Number","Month"])["Net_Sales"].sum().sort_index()

plt.figure(figsize=(4,2))
plt.plot(monthly_sales.index.get_level_values("Month"),monthly_sales.values,marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
#Quantity vs Net Sales: Scatter Plot        

plt.figure(figsize=(4,2))

plt.scatter(df["Quantity"], df["Net_Sales"])
plt.title("Quantity vs Net Sales")
plt.xlabel("Quantity")
plt.ylabel("Net Sales")

plt.show()
```

```python
#Net Sales Distribution: Histogram
plt.figure(figsize=(6,3))

plt.hist(df["Net_Sales"], bins=20)

plt.title("Net Sales Distribution")
plt.xlabel("Net Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
#Payment Method Distribution: Pie Chart
payment_counts = df["Payment_Method"].value_counts()

plt.figure(figsize=(6,3))
plt.pie(payment_counts.values,labels=payment_counts.index,autopct="%1.1f%%")

plt.title("Payment Method Distribution")

plt.show()
```

```python
#Category-wise Orders: Count Plot
plt.figure(figsize=(5,2))

sns.countplot(data=df, x="Category")

plt.title("Category-wise Orders")
plt.xlabel("Category")
plt.ylabel("Number of Orders")

plt.show()
```

```python
#Category-wise Sales: Bar Plot
category_sales = df.groupby("Category")["Net_Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(6, 2))

sns.barplot(x=category_sales.index, y=category_sales.values)

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Net Sales")

plt.show()
```

```python
#Sales Distribution by Category: Box Plot
plt.figure(figsize=(6,2))

sns.boxplot(data=df, x="Category", y="Net_Sales")

plt.title("Sales Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Net Sales")

plt.show()
```

```python
#Quantity vs Net Sales by Category
plt.figure(figsize=(6, 3))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Net_Sales",
    hue="Category"
)

plt.title("Quantity vs Net Sales by Category")
plt.xlabel("Quantity")
plt.ylabel("Net Sales")

plt.show()
```

```python
#Numerical Correlation Heatmap
plt.figure(figsize=(7,3))

correlation = df.select_dtypes(include=np.number).corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm",fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```

```python
#Quantity, Unit Price and Net Sales – Pair Plot.
sns.pairplot(
    df[["Quantity", "Unit_Price", "Net_Sales"]]
)
plt.figure(figsize=(7,3))
plt.show()
```

```python
# Discount vs Net Sales

discount_sales = df.groupby("Discount_Percent")["Net_Sales"].mean()

print("Average Net Sales by Discount:")
print(discount_sales)

print("\nCorrelation between Discount and Net Sales:")
print(df["Discount_Percent"].corr(df["Net_Sales"]))
```
